```python
import pandas as pd

customers_data = {
    "CustomerID": [101, 102, 103, 104, 105],
    "Name": ["Aisha", "Rahul", "Sara", "John", "Meera"],
    "City": ["Bangalore", "Mysore", "Bangalore", "Delhi", "Chennai"]
}

customers = pd.DataFrame(customers_data)

customers.to_csv("customers.csv", index=False)

print("customers.csv created successfully!")
```

```python
sales_data = {
    "SaleID": [1, 2, 3, 4, 5, 6, 7, 8],
    "CustomerID": [101, 102, 101, 103, 105, 102, 103, 105],
    "Product": [
        "Laptop", "Phone", "Mouse", "Keyboard",
        "Monitor", "Headphones", "Tablet", "Mouse"
    ],
    "Category": [
        "Electronics", "Electronics", "Accessories",
        "Accessories", "Electronics", "Accessories",
        "Electronics", "Accessories"
    ],
    "Amount": [60000, 20000, 1000, 2500, 15000, 3000, 25000, 800]
}

sales = pd.DataFrame(sales_data)

sales.to_csv("sales.csv", index=False)

print("sales.csv created successfully!")
```

```python
import pandas as pd

customers = pd.read_csv("customers.csv")
sales = pd.read_csv("sales.csv")
print(customers)

print(sales)
print("Customer dataset:", customers.shape)
print("Sales dataset:", sales.shape)
```

```python
df = pd.merge(
    sales,
    customers,
    on="CustomerID",
    how="left",
    validate="many_to_one"
)

print("\nCOMBINED SALES DATA")
print(df)
```

```python
print("\nOVERALL BUSINESS SUMMARY")
print("Total revenue:", df["Amount"].sum())
print("Number of sales:", df["SaleID"].nunique())
print("Unique customers:", df["CustomerID"].nunique())
print("Average sale:", round(df["Amount"].mean(), 2))
```

```python
city_summary = df.groupby("City").agg(
    Total_Revenue=("Amount", "sum"),
    Number_of_Sales=("SaleID", "nunique"),
    Unique_Customers=("CustomerID", "nunique"),
    Average_Sale=("Amount", "mean")
).reset_index()

city_summary = city_summary.sort_values(
    "Total_Revenue", ascending=False
)

print("\nCITY PERFORMANCE")
print(city_summary)
```

```python
category_summary = df.groupby("Category").agg(
    Total_Revenue=("Amount", "sum"),
    Number_of_Sales=("SaleID", "nunique"),
    Average_Sale=("Amount", "mean")
).reset_index()

category_summary = category_summary.sort_values(
    "Total_Revenue", ascending=False
)

print("\nCATEGORY PERFORMANCE")
print(category_summary)
```

```python
product_summary = df.groupby("Product").agg(
    Total_Revenue=("Amount", "sum"),
    Number_of_Sales=("SaleID", "nunique")
).reset_index()

product_summary = product_summary.sort_values(
    "Total_Revenue", ascending=False
)

print("\nPRODUCT PERFORMANCE")
print(product_summary)
```

```python
df.to_csv("merged_sales_data.csv", index=False)
city_summary.to_csv("city_summary.csv", index=False)
category_summary.to_csv("category_summary.csv", index=False)
product_summary.to_csv("product_summary.csv", index=False)

print("\nAll reports exported successfully.")

```
